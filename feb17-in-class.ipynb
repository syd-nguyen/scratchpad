```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/metabric.csv')
df.head()
```

```python
df.columns
```

```python
vars = ['Age at Diagnosis', 'Type of Breast Surgery', 'Cancer Type',
       'Chemotherapy', 'Hormone Therapy', 'Lymph nodes examined positive',
       'Mutation Count', 'Nottingham prognostic index',
       'Overall Survival (Months)', 'Overall Survival Status', 'Radio Therapy',
       'TMB (nonsynonymous)', 'Tumor Size', 'Tumor Stage']

vars_num = ['Age at Diagnosis', 'Tumor Size', 'Mutation Count', 'Nottingham prognostic index', 'TMB (nonsynonymous)']
stages = pd.get_dummies(df['Tumor Stage'], dtype=int, drop_first=True, prefix='stage')
```

```python
df = pd.concat([df, stages])
df.head()
```

```python
y = df['Overall Survival (Months)']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df,y, # Feature and target variables
                                                    test_size=.2, # Split the sample 80 train/ 20 test
                                                    random_state=100) # For replication purposes
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model = model.fit(X_train, y_train)
y_hat = model.predict(X_test)
```
